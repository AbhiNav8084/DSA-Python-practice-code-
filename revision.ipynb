```python
s = "i am good"
s= list(s)
n= len(s)

def preserveSTRspace():
    left = 0
    right = n-1

    while left < right:

        if s[left] == " ":
            left += 1
        
        elif s[right] == " ":
            right -= 1

        else:
            s[left],s[right] = s[right],s[left]
            left += 1
            right -= 1
    return ''.join(s)
preserveSTRspace()
```

```python
s = "anagram" 
t = "nagaram"

def anagramFREQ():

    if len(s) != len(t):
        return False

    count = [0]*26

    for i in s:
        count[ord(i) - ord('a')] += 1
    
    for i in t:
        count[ord(i) - ord('a')] -= 1

    for i in count:
        if count[i] != 0:
            return False
    return True
anagramFREQ()
```
