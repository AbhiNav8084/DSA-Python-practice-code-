```python
n=8
def perfectNum(n):
    sum = 0
    for i in range(1,n):
        if n % i == 0:
            sum = sum + i
    return sum == n
perfectNum(n)
```

```python
n =9
def primeNum(n):
    if n < 2 :
        return False
    for i in range(2, n):
        if n % i == 0:
            return False
    return True
primeNum(n)
```

```python
n=46
def isPrime(n):
    if n < 2 :
        return False
    for i in range(2, n):
        if n % i == 0:
            return False
    return True
def tillN(n):
    count = 0
    for i in range (2,n):
        if isPrime(i):
            count = count +1
    return count 
tillN(n)
```

```python
arr = [1,2,2,3,4,5,6]
n = len(arr)
def ascOrd(arr,n):
    for i in range(n-1):
        if arr[i] > arr[i+1]:
            return False
    return True
ascOrd(arr,n)
```

```python
arr = [1,2,2,3,4,5,6]
n= len(arr)
def ascendingOrd(arr,n):
    for i in range(n-1):
        for j in range(i+1,n):
            if arr[i] > arr[j]:
                return False
    return True
ascendingOrd(arr,n)
```

```python
arr = [1,2,3,4,5,6]
n = len(arr)
def revArr(arr,n):
    temp_arr = [0]*n
    for i in range(n):
        temp_arr[n-i-1] = arr[i]
    for i in range(n):
        arr[i] = temp_arr[i]
    return arr
revArr(arr,n)
```

```python
arr = [1,2,3,4,5,6]
n = len(arr)
def rev_arr(arr,n):
    left = 0
    right = n-1
    while left < right:
        arr[left],arr[right] =arr[right],arr[left]
        left = left + 1
        right = right -1
    return arr
rev_arr(arr,n)
```

```python
arr = [1,2,3,4,5,6]
n = len(arr)
def arr_rev(arr,n):
    for i in range(n//2):
        arr[i],arr[n-i-1] = arr[n-i-1],arr[i]
    return arr
arr_rev(arr,n)
```
