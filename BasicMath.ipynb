## 1 total number of digits in number

```python

n=178604
def countDigit(n):
    if n == 0:
        return 1
    count = 0
    while n > 0 :
        count = count+1
        n= n // 10
    return count
countDigit(n)
```

## 2 total number of odd digits in number

```python

n= 22
def oddDigit(n):
    count = 0
    while n > 0:
        lastDigit = n % 10
        if lastDigit % 2 != 0:
            count = count+1
        n= n//10
    return count
oddDigit(n)
```

## 3 return reverse number

```python

n=123456
def reverseNumber(n):
    revNum = 0 
    while n > 0 :
        lastDigit = n % 10
        revNum = revNum*10 + lastDigit
        n = n //10
    return revNum
reverseNumber(n)
```

## 4 check whether the number is palindrome number or not

```python

n=4331
def isPalindrome(n):
    copy= n
    revNum = 0
    while n > 0 :
        lastDigit =  n % 10 
        revNum = revNum * 10 + lastDigit
        n = n // 10
    return revNum == copy
isPalindrome(n)
```

## 5 return the largest digit from number

```python

n=3061
def largestDigit(n):
    largestNum = 0
    while n > 0:
        lastDigit = n % 10 
        if lastDigit > largestNum:
            largestNum = lastDigit
        n = n // 10
    return largestNum
largestDigit(n)
```

## 6 Factorial of the given number 

```python

n = 4
def Factorial(n):
    if n == 0:
        return  1
    fact = 1
    for i in range (1, n+1):
        fact = fact * i
    return fact
Factorial(n)
```

## 7 Check if the Number is Armstrong

```python

n= 153
def countDigit(n):
    if n == 0 :
        return 1
    count = 0
    while n > 0:
        count = count + 1
        n= n // 10
    return count
countDigit(n)
def isArmstrong(n):
    digit = countDigit(n)
    copy= n
    sum = 0
    while n > 0:
        lastDigit = n % 10
        sum  = sum + pow(lastDigit , digit )
        n = n//10
    if sum == copy:
        return True
    else:
        return False
isArmstrong(n)
```

## 8 find total digits of prime number in give number

```python

n= 41365
def PrimeDigit(n):
    count = 0
    while n > 0:
        lastDigit = n % 10

        if lastDigit % 2 != 0:
            count = count+1
        elif lastDigit == 2 :
            count = count + 1
        n= n//10
    return count
PrimeDigit(n)
```
