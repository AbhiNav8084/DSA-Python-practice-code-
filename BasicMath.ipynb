## 1 total number of digits in number

```python

n=17860
def countDigit(n):
    if n == 0:
        return 1
    count = 0
    while n > 0 :
        count = count+1
        n= n // 10
    return count
countDigit(n)
```

## 2 total number of odd digits in number

```python

n= 122
def oddDigit(n):
    count = 0
    while n > 0:
        lastDigit = n % 10
        if lastDigit % 2 != 0:
            count = count+1
        n= n//10
    return count
oddDigit(n)
```

## 3 return reverse number

```python

n=123456
def reverseNumber(n):
    revNum = 0 
    while n > 0 :
        lastDigit = n % 10
        revNum = revNum*10 + lastDigit
        n = n //10
    return revNum
reverseNumber(n)
```

## 4 check whether the number is palindrome number or not

```python

n=4331
def isPalindrome(n):
    copy= n
    revNum = 0
    while n > 0 :
        lastDigit =  n % 10 
        revNum = revNum * 10 + lastDigit
        n = n // 10
    return revNum == copy
isPalindrome(n)
```

## 5 return the largest digit from number

```python

n=3061
def largestDigit(n):
    largestNum = 0
    while n > 0:
        lastDigit = n % 10 
        if lastDigit > largestNum:
            largestNum = lastDigit
        n = n // 10
    return largestNum
largestDigit(n)
```

## 7 Check if the Number is Armstrong
#### (METHOD-1)

```python

n= 153
def countDigit(n):
    if n == 0 :
        return 1
    count = 0
    while n > 0:
        count = count + 1
        n= n // 10
    return count
countDigit(n)
def isArmstrong(n):
    digit = countDigit(n)
    copy= n
    sum = 0
    while n > 0:
        lastDigit = n % 10
        sum  = sum + pow(lastDigit , digit )
        n = n//10
    if sum == copy:
        return True
    else:
        return False
isArmstrong(n)
```

## 10 Check for Perfect Number
#### (METHOD-1)

```python
n= 496
def perfectNumber(n):
    sum = 0
    for i in range (1,n-1):
        if n % i == 0:
            sum = sum + i
    return sum == n
perfectNumber(n)
```

## 12 Count Prime Numnber till N

```python
n =73
def isPrime(n):
    if n < 2 :
        return False
    for i in range(2,n):
        if n % i == 0:
            return False
    return True
isPrime(n)

def PrimeTillN(n):
    count = 0
    for i in range (2,n+1):
        if isPrime(i):
            count = count +1
    return count
PrimeTillN(n)
```

## 13 GCD of given Numbers
#### (METHOD -1)

```python
n1=12
n2 =16
def GCD(n1,n2):
    largest_gcd = 1
    for i in range (1,min(n1,n2)+1):
        if (n1 % i == 0 and n2 % i == 0):
            largest_gcd = i
    return largest_gcd
GCD(n1,n2)
```

## 14 LCM of given Numbers
#### (METHOD -1)

```python
n1=12
n2=16
def GCD(n1,n2):
    while(n1 != 0 and n2 != 0):
        if n1 > n2 :
            n1 = n1- n2
        else :
            n2 = n2 - n1
    if n2 == 0 : return n1
    return n2
GCD(n1,n2)
def LCM(n1,n2):
    gcd=GCD(n1,n2)
    lcm = (n1*n2//gcd)
    return lcm
LCM(n1,n2)
```

## 14 LCM of given Numbers
#### (METHOD -3)

```python
n1=12
n2=16
def LCM(n1,n2):
    lcm =0 
    n = max(n1,n2)
    i =1
    while True:
        multiple = n*i
        if multiple % n1 == 0 and multiple % n2 == 0:
            lcm = multiple
            break
        i += 1
    return lcm
LCM(n1,n2)
```
