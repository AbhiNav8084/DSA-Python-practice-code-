```python
# 1 total number of digits in number
n=178694
def countDigit(n):
    if n == 0:
        return 1
    count = 0
    while n > 0 :
        count = count+1
        n= n // 10
    return count
countDigit(n)
```

```python
# 2 total number of odd digits in number
n= 22
def oddDigit(n):
    count = 0
    while n > 0:
        lastDigit = n % 10
        if lastDigit % 2 != 0:
            count = count+1
        n= n//10
    return count
oddDigit(n)
```

```python
# 3 return reverse number
n=123456
def reverseNumber(n):
    revNum = 0 
    while n > 0 :
        lastDigit = n % 10
        revNum = revNum*10 + lastDigit
        n = n //10
    return revNum
reverseNumber(n)
```

```python
#4 check whether the number is palindrome number or not
n=4334
def isPalindrome(n):
    copy= n
    revNum = 0
    while n > 0 :
        lastDigit =  n % 10 
        revNum = revNum * 10 + lastDigit
        n = n // 10
    return revNum == copy
isPalindrome(n)
```

```python
#5 return the largest digit from number
n=3061
def largestDigit(n):
    largestNum = 0
    while n > 0:
        lastDigit = n % 10 
        if lastDigit > largestNum:
            largestNum = lastDigit
        n = n // 10
    return largestNum
largestDigit(n)
```
