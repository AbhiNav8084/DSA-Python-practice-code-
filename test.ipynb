```python
s = "Things To Remember"
s = list(s)
n = len(s)

def reverseStr():
    left = 0
    right = n-1

    while left < right:

        if s[left] == " ":
            left += 1
        elif s[right] == " ":
            right -= 1
        else:
            s[left],s[right] = s[right],s[left]
            left += 1
            right -= 1
    return ''.join(s)
reverseStr()
```

```python
s = "hannah"
n = len(s)

def palindromeCheck():
    left = 0
    right = n-1

    while left < right:
        if s[left] != s[right]:
            return False
        left += 1
        right -= 1
    return True
palindromeCheck()
```

```python
s = "0214638"
n = len(s)

def largeOddNum():
    index = -1
    for i in range(n-1,-1,-1):
        if (int(s[i]) % 2) == 1:
            index = i 
            break
    i = 0
    while i <= index and s[i] == '0':
        i += 1
    return s[ i: index + 1 ]
largeOddNum()
```

```python
s = "paper"
t = "title"

def isomorphicString():
    a1 = [0]*256
    a2 = [0]*256

    n = len(s)

    for i in range(n):
        if a1[ord(s[i])] != a2[ord(t[i])]:
            return False
        a1[ord(s[i])] = i + 1
        a2[ord(t[i])] = i + 1
    return True
isomorphicString()
```

```python
s = "abcde"
goal = "adeac"

def rotateString():
    if len(s) != len(goal):
        return False

    for i in range(len(s)):
        if s[i:] + s[:i] == goal:
            return True
    return False
rotateString()
```

```python
s = "eat"
t = "tea"

def anagramStrings():

    if len(s) != len(t):
        return False

    count= {}

    for i in s:
        if i in count:
            count[i] += 1
        else:
            count[i] =1

    for i in t:
        if i in count:
            count[i] -= 1
        else:
            return False

    for i in count:
        if count[i] != 0:
            return False
    return True

anagramStrings()
```
