## 1 Highest Occurring Element in an Array

```python
nums = [1,2,2,3,3,3]
n=len(nums)
def MostFrequentElem():
    freq ={}
    for x in nums :
        if x in freq :
            freq[x] += 1
        else :
            freq[x] = 1
    maxfreq =0
    maxElem = -1

    for x in freq:
        if freq[x] > maxfreq:
            maxfreq = freq[x]
            maxElem = x
        elif freq[x] == maxfreq:
            maxElem = min(maxElem,x)
    return maxElem
MostFrequentElem()
```

## 2 Second Highest Occurring Element

```python
nums =[10, 9 ,7, 7]
n=len(nums)
def secondMostFrequentElement():
    freq ={}
    for i in nums:
        if i in freq:
            freq[i] +=1
        else :
            freq[i] =1
    maxFreq = 0
    maxElem = 0
    SecondMaxFreq = 0
    SecondMaxElem = 0

    for i in freq :
        if freq[i] > maxFreq:
            SecondMaxFreq = maxFreq
            SecondMaxElem = maxElem
            maxFreq = freq[i]
            maxElem = i
        elif freq[i] == maxFreq:
            maxElem = min(maxElem,i)
        elif freq[i] > SecondMaxFreq and freq[i] < maxFreq:
            SecondMaxFreq = freq[i]
            SecondMaxElem = i
        elif freq[i] == SecondMaxFreq:
            SecondMaxElem = min(SecondMaxElem,i)
    return SecondMaxElem
secondMostFrequentElement()
```

## 3 Sum of Highest and Lowest Frequency

```python
arr = [1, 2, 2, 6,6,6,6]
n  = len(arr)
def SumMaxFrqMinFrq():
    mpp = {}
    maxFreq = 0
    minFreq = n

    for i in arr:
        if i in mpp:
            mpp[i] += 1
        else:
            mpp[i] = 1
    for freq in mpp.values():
        maxFreq = max(maxFreq, freq)
        minFreq = min(minFreq, freq)
    return maxFreq + minFreq
SumMaxFrqMinFrq()
```
