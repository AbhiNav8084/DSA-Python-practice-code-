## 1 Reverse a String 

```python
#s = list('hello')
s = ["h", "e" ,"l" ,"l" ,"o"]
n = len(s)
def revStr():
    left = 0
    right = n-1
    while left < right:
        s[left] ,s[right] = s[right], s[left]
        left = left +1
        right = right -1
    return s
revStr()
```

## 2 Palindrome Check

```python
s = "hannah"
def isPalindrome():
    left  =0 
    right = len(s)-1
    while left < right :
        if s[left] != s[right]:
            return False
        left = left + 1
        right = right -1
    return True
isPalindrome()
```
