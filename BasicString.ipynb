## 1 Reverse a String 

```python
#s = list('hello')
s = ["h", "e" ,"l" ,"l" ,"o"]
n = len(s)
def revStr():
    left = 0
    right = n-1
    while left < right:
        s[left] ,s[right] = s[right], s[left]
        left = left +1
        right = right -1
    return s
revStr()
```

## 2 Palindrome Check

```python
s = "hannah"
def isPalindrome():
    left  =0 
    right = len(s)-1
    while left < right :
        if s[left] != s[right]:
            return False
        left = left + 1
        right = right -1
    return True
isPalindrome()
```

## 3 Largest Odd Number in a String
### (that is a substring)

```python
s = "5347"
n = len(s)
def lrgOdNu():
    ind = -1
    for i in range(n-1,-1,-1):
        if (int(s[i])%2) == 1:
            ind = i
            break
    i = 0
    while i <= ind and s[i] == '0':
        i += 1
    return s[i:ind + 1]
lrgOdNu()
```

## 4 Longest Common Prefix

#### Method-1


```python
s = ["flowers" , "flow" , "fly", "flight" ]
def lcp():
    if not s:
        return ""
    s.sort()
    first = s[0]
    last = s[-1]
    prefix = ""

    for i in range(min(len(first),len(last))):
        if first[i] != last[i]:
            break
        prefix += first[i]
    return prefix
lcp()
```

```python
s = ["flowers" , "flow" , "fly", "flight" ]
def lcp():
    if not s:
        return ""
    s.sort()
    first = s[0]
    last = s[-1]
    prefix = []

    for i in range(min(len(first),len(last))):
        if first[i] != last[i]:
            return ''.join(prefix)
        prefix.append(first[i])

    return ''.join(prefix)
lcp()
```

## 5 Isomorphic String

```python
s = "egg" 
t = "add"
n = len(s)

def IsoStr():
    a1 = [0]*256
    a2 = [0]*256

    for i in range(n):
        if a1[ord(s[i])] != a2[ord(t[i])]:
            return False
        a1[ord(s[i])] = i + 1
        a2[ord(t[i])] = i + 1
    return True
IsoStr()
```

## 6 Rotate String

#### Method-1(Brute Force)

```python
s = "abcde"
goal = "cdeab"

def RotateSTR():
    n = len(s)
    if n != len(goal):
        return False
    
    for i in range(n):
        rotated = s[i:] + s[:i]
        if rotated == goal:
            return True
    return False
RotateSTR()
```

#### Method-2(Using Slicing)

```python
s = "abcde"
goal = "cdeab"

def RotateStr():
    if len(s) != len(goal):
        return False
    doubled_s = s + s
    return goal in doubled_s
RotateStr()
```

#### Method-3(Using Slicing)

```python
s = "abcde"
goal = "cdeab"

def RotateStR():
    if len(s) != len(goal):
        return False

    for i in range(len(s)):
        if s[i:] + s[:i] == goal:
            return True
    return False
RotateStR()
```
