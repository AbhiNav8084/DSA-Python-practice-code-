## 1 Reverse a String 
#### METHOD 1

```python
# s = list('hello')
s = ["h", "e" ,"l" ,"l" ,"o"]
n = len(s)
def revStr():
    left = 0
    right = n-1
    while left < right:
        s[left] ,s[right] = s[right], s[left]
        left = left +1
        right = right -1
    return ''.join(s)
revStr()
```

#### METHOD 2

```python
s = "HELLO"
n= len(s)
def strRev():
    rev = ""
    for i in range(n):
        rev = s[i]+rev
    return rev
strRev()
```

## 2 Palindrome Check

```python
s = "hannah"
def isPalindrome():
    left  =0 
    right = len(s)-1
    while left < right :
        if s[left] != s[right]:
            return False
        left = left + 1
        right = right -1
    return True
isPalindrome()
```

## 3 Largest Odd Number in a String
### (that is a substring)

```python
s = "5347"
n = len(s)
def lrgOdNu():
    ind = -1
    for i in range(n-1,-1,-1):
        if (int(s[i])%2) == 1:
            ind = i
            break
    i = 0
    while i <= ind and s[i] == '0':
        i += 1
    return s[i:ind + 1]
lrgOdNu()
```

## 4 Longest Common Prefix

#### Method-1


```python
s = ["flowers" , "flow" , "fly", "flight" ]
def lcp():
    if not s:
        return ""
    s.sort()
    first = s[0]
    last = s[-1]
    prefix = ""

    for i in range(min(len(first),len(last))):
        if first[i] != last[i]:
            break
        prefix += first[i]
    return prefix
lcp()
```

## 5 Isomorphic String

```python
s = "egg" 
t = "add"
n = len(s)

def IsoStr():
    a1 = [0]*256
    a2 = [0]*256

    for i in range(n):
        if a1[ord(s[i])] != a2[ord(t[i])]:
            return False
        a1[ord(s[i])] = i + 1
        a2[ord(t[i])] = i + 1
    return True
IsoStr()
```

## 6 Rotate String

#### Method-1(Brute Force)

```python
s = "abcde"
goal = "cdeab"

def RotateSTR():
    n = len(s)
    if n != len(goal):
        return False
    
    for i in range(n):
        rotated = s[i:] + s[:i]
        if rotated == goal:
            return True
    return False
RotateSTR()
```

#### Method-2(Using Slicing)

```python
s = "abcde"
goal = "cdeab"

def RotateStr():
    if len(s) != len(goal):
        return False
    doubled_s = s + s
    return goal in doubled_s
RotateStr()
```

#### Method-3(Using Slicing)

```python
s = "abcde"
goal = "cdeab"

def RotateStR():
    if len(s) != len(goal):
        return False

    for i in range(len(s)):
        if s[i:] + s[:i] == goal:
            return True
    return False
RotateStR()
```

## 7 Valid Anagram

#### Method-1(Brute Force)

```python
s = "anagram" 
t = "nagaram"

def ValidAnagram():
    if len(s) != len(t):
        return False
    
    return sorted(s) == sorted(t)
ValidAnagram()
```

#### method-2 (Frequency Array)

```python
s = "anagram" 
t = "nagaram"

def ValiDAnaGram():
    if len(s) != len(t):
        return False
    
    count = [0]*26

    for c in s:
        count[ord(c) - ord('a')] += 1
    
    for c in t:
        count[ord(c) - ord('a')] -= 1
    
    for i in count:
        if i != 0:
            return False
    return True

ValiDAnaGram() 
```

#### method-3 (Hash Map)

```python
s = "anagram" 
t = "nagaram"

def ValidAnaGRAM():
    if len(s) != len(t):
        return False

    count = {}

    for c in s:
        if c in count:
            count[c] += 1
        else:
            count[c] =1

    for c in t:
        if c in count :
            count[c] -= 1
        else:
            return False

    for c in count:
        if count[c] != 0:
            return False
    return True

ValidAnaGRAM()
```

## 8 Reverse String by preserving spaces

#### using 2 pointer approach

```python
s = "i am good"
s = list(s)
n = len(s)

def preserveSpace():
    left = 0
    right = n-1

    while left < right :
        if s[left] == " ":
            left += 1

        elif s[right] == " ":
            right -= 1

        else:
            s[left],s[right] = s[right],s[left]
            left += 1
            right -= 1
    return ''.join(s)
preserveSpace()
```

## 9 Sort Characters by Frequency

#### Method-1 (Using Hash Map)

```python
s = "tree"
def frequencySort():

    freq = {}
    for i in s:
        if i in freq:
            freq[i] += 1
        else:
            freq[i] = 1

    chars = list(freq.keys())

    chars.sort(key=lambda i: (-freq[i],i))

    return chars

frequencySort()
```

## 10 Count Of Vowels
#### Method-1 (Brute Force)

```python
str = "kodnest"
n = len(str)

def VowelCount():
    count = 0
    for i in range(n):
        if str[i] in ('a','e','i','o','u'):
            count += 1
    return count
VowelCount()
```

#### METHOD 2(optimal approach)

```python

s = "kodnest"

def vowelCount(s):
    count = 0
    vowels = "aeiou"

    for i in s.lower():
        if i in vowels:
            count += 1

    return count

vowelCount(s)
```

## 11 Count Target String in Source String

```python
s = "nayana"
n = len(s)
def countTarget():
    target = 'a'
    count = 0
    for i in range(n):
        if s[i] == target:
            count += 1
    return count 
countTarget()
```
