```python
n=5
spaces = 2*n-2
for i in range(1,2*n):
    stars = i
    if i > n:
        stars = 2*n-i
    print("*"*stars,end="")
    print(" "*spaces,end="")
    print("*"*stars,end="")
    print()
    if i < n:
        spaces -=2
    else:
        spaces+=2
```
