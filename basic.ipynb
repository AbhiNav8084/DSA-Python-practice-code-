## 1 Leap Year

```python
year = 2024

if year % 400 == 0:
    print("Leap year")
elif year % 100 == 0:
    print("Not a leap year")
elif year % 4 == 0:
    print("Leap year")
else:
    print("Not a leap year")
```

## 2 powerOf — raise to a power
#### method-1

```python
def powerOf(base,exp):
    result = 1
    for i in range(exp):
        result = result * base
    return result
powerOf(3,3) 
```

#### method-2

```python
def powerOF():
    base = int(input("Enter base"))
    exp = int(input("Enter exp"))
    res =1
    for i in range(exp):
        res *= base
    return res 
powerOF()
```
