## 2 Sum all the elements in array
#### (METHOD-1)

```python
arr = [1,2,3,4,5]
sum = 0
for x in arr:
    sum = sum + x
print(sum)
```

## 2 Sum all the elements in array
#### (METHOD-2)

```python
arr = [1,2,3,4,5,6]
sum =0 
for i in range(len(arr)):
    sum = sum + arr[i]
print(sum)
```

## 3 Count total odd digit in array
#### (METHOD-1)

```python
arr = [3,5,3,6,8]
count =0 
for i in arr:
    lastDigit = i % 10
    if lastDigit % 2 != 0:
        count = count +1
print(count)
```

## 3 Count total odd digit in array
#### (METHOD-2)

```python
arr = [2,4,1,55,7,9]
count= 0
for i in arr :
    if i % 2 != 0:
        count += 1
print(count)
```
