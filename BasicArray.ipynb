## 4 Check if the Array is Sorted 
#### (METHOD-1)

```python
arr = [1,2,2,3,4,5,6]
n=len(arr)
def ascendingArr(arr,n):
    for i in range(n-1):
        for j in range(i+1,n):
            if arr[i] > arr[j]:
                return False
    return True
ascendingArr(arr,n)
```

## 4 Check if the Array is Sorted

```python
arr = [1,2,2,3,4,5,6]
n=len(arr)
def ascendingArr(arr,n):
    for i in range(n-1):
        if arr[i]>arr[i+1]:
            return False
    return True
ascendingArr(arr,n)
```

## 5 Reverse an array
#### (METHOD-1)

```python
arr = [1,2,3,4,5,6]
n= len(arr)
def Reversed_arr(arr,n):
    temp_arr = [0]*n
    for i in range (n):
        temp_arr[n-i-1]=arr[i]
    for i in range(n):
        arr[i] = temp_arr[i]
    return arr
Reversed_arr(arr,n)
```

## 5 Reverse an array
#### (METHOD-2)

```python
arr = [1,2,3,4,5,6]
n= len(arr)
def Reversed_arr(arr,n):
    for i in range(n//2):
        arr[i],arr[n-i-1] = arr[n-i-1],arr[i]
    return arr
Reversed_arr(arr,n)
```

## 5 Reverse an array
#### (METHOD-3) 
##### Two Pointer

```python
arr = [11,2,3,4,5,6]
n= len(arr)
def Reversed_arr(arr,n):
    left = 0 
    right = n-1
    while left < right :
        arr[left],arr[right] = arr[right],arr[left]
        left =left +1
        right = right -1
    return arr
Reversed_arr(arr,n)
```

## 6 Find Maximum and Minimum in array
#### (METHOD-1)

```python
arr=[8,2,56,63,2,574,6]
n=len(arr)
def max_min(arr):
    max=arr[0]
    min=arr[0]
    for i in range (n):
        if arr[i] > max:
            max=arr[i]
        elif arr[i] < min:
            min =arr[i]
    print("Maximum :",max , ", Minimum :",min)
max_min(arr)
```

## 7 countAbove — count values over a limit
#### METHOD-1(limit as a variable)

```python
arr = [20,30,40,50,60,70,80]
n = len(arr)
def CountAbove(arr,n):
    limit = 50
    count = 0
    for i in range(n):
        if arr[i] > limit :
            count = count + 1
    return count
CountAbove(arr,n)
```

#### METHOD-2(limit as a function parameter)

```python
arr = [20,30,40,50,60,70,80]
n = len(arr)
def CountAbove(arr,n,limit):
    count = 0
    for i in range(n):
        if arr[i] > limit :
            count = count + 1
    return count
CountAbove(arr,n,50)
```
